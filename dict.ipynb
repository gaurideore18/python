```python
# how to reverse string data using for loop
string='software'
list=string.split()
result=[]
for i in list:
    if i not in result:
        result.append(i[::-1])
output=''.join(result)
print(output)
```

```python
string ='bangalore'
dict={}
output=''
for ch in string:
    dict[ch]=dict.get(ch,0)+1
for key,value in dict.items():
    output=output+k+str(v)
print(output)
```
